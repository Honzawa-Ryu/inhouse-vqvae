```python
import pathlib as Path
import sys
sys.path.append(str(Path.Path.cwd().parent.resolve()))

import lpips
import src.lpips_loss
from PIL import Image
import torchvision
import torch
```

```python
p = Image.open('/workspace/inhouse-vqvae/VQVAE/data/hist_class/mouse_brain/HE_brain_RO555_512_9728.png')
t = torchvision.transforms.functional.to_tensor(p)
p = Image.open('/workspace/inhouse-vqvae/VQVAE/data/hist_class/mouse_brain/HE_brain_RO2050_26112_3072.png')
s = torchvision.transforms.functional.to_tensor(p)
```

```python
lpips_loss_fn = lpips.LPIPS(net='vgg16', version='0.1').to('cuda')
inhouse_loss = src.lpips_loss.lpips_loss().to('cuda')
```

```python
loss_a = lpips_loss_fn(t.unsqueeze(0), s.unsqueeze(0))
loss_b = inhouse_loss(t.unsqueeze(0), s.unsqueeze(0))
```

```python
print(f'LPIPS loss: {loss_a.item()}')
print(f'In-house LPIPS loss: {loss_b.item()}')
print(f'Difference: {abs(loss_b.item() / loss_a.item())}')
```
